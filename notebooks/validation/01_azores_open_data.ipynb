## Analytical boundary

The Azores files are public raw/source data and are therefore handled differently from restricted CETI recordings. The public repository points to the source and provides provenance-aware retrieval, but it does not mirror the archive. Publication figures are generated by the tested `validation.azores` extraction and analysis module after checksum validation; numerical reproduction remains distinct from merely loading the source files. The manuscript choices exposed by that reproduction are tracked in `docs/known_issues.md`.

This notebook does not claim that the Azores files are CETI data and does not combine them with restricted CETI material. See `docs/validation_pipeline.md`.
