# M6 tactic-search coupling and phase check

Read `README.md` in this folder first. Part 1 checks every number the letter quotes for this module against the result files (a MISMATCH means the letter, the file or the code drifted). Part 2 shows the evidence.

```python
import sys, pathlib
import pandas as pd, matplotlib.pyplot as plt
for p in (pathlib.Path.cwd(), *pathlib.Path.cwd().parents):
    if (p / "review_r1" / "common.py").exists():
        sys.path.insert(0, str(p / "review_r1")); break
from common import load_claims, observed, matches, RESULTS
pd.set_option("display.width", 200); pd.set_option("display.max_colwidth", 70)
```

## 1. Letter vs result files

```python
rows = []
for c in load_claims("M6"):
    if c["status"] == "printed_only":
        rows.append((c["id"], c["status"], c["letter"], None, "not checked", c["text"])); continue
    v = observed(c); rows.append((c["id"], c["status"], c["letter"], round(v, 4), "OK" if matches(c, v) else "MISMATCH", c["text"]))
pd.DataFrame(rows, columns=["claim", "status", "letter", "observed", "check", "text"])
```

## 2. Evidence

```python
import json
F = json.load(open(RESULTS/"foraging_models.json"))["dswp"]
pd.DataFrame({"all buzzes": [F["D4_terminal_matched"]["is_b"]["odds_ratio"]],
              "bottom phase only": [F["P_D4_terminal_bottom_only"]["is_b"]["odds_ratio"]]},
             index=["OR: Type B is dive's last buzz"]).T.plot.bar(); plt.axhline(1, c="k", lw=.5)
display(pd.read_csv(RESULTS/"transit_buzz_profile.csv").round(2))
```
